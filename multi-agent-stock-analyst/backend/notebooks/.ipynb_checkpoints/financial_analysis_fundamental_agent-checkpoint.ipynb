```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option("display.float_format", "{:,.2f}".format)
```

```python
df = pd.read_csv(r"C:\Users\haran\Documents\Infosys Springboard Internship\-AI-Powered-Multi-Agent-System-for-Market-Analysis-and-Research\multi-agent-stock-analyst\backend\Data\cleaned\cleaned_nse_top100.csv")

df["date"] = pd.to_datetime(df["date"])

df = df.sort_values(["ticker", "date"])

print("Dataset shape:", df.shape)
print("Total stocks:", df["ticker"].nunique())

df.head()
```

```python
df["year"] = df["date"].dt.year

annual_analysis = df.groupby(
    ["ticker", "year"]
).agg(
    annual_turnover=("turnover", "sum"),
    average_close=("close", "mean"),
    first_close=("close", "first"),
    last_close=("close", "last"),
    trading_days=("date", "count")
).reset_index()

annual_analysis.head()
```
